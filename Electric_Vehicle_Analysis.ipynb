```python
import pandas as pd

df = pd.read_csv("Electric_Vehicle_Population_Data.csv")

print("Rows and columns:", df.shape)

df.head()
```

```python
# Step 2: Inspect the dataset

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Step 3: Check for duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Step 4: Clean important text columns

text_columns = [
    "County",
    "City",
    "State",
    "Make",
    "Model",
    "Electric Vehicle Type",
    "Clean Alternative Fuel Vehicle (CAFV) Eligibility",
    "Electric Utility"
]

# Remove extra spaces
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Standardize Make and Model
df["Make"] = df["Make"].str.upper()
df["Model"] = df["Model"].str.upper()

# Fill missing Make/Model
df["Make"] = df["Make"].fillna("UNKNOWN")
df["Model"] = df["Model"].fillna("UNKNOWN")

print("Text cleaning completed.")
```

```python
# Step 5: Create simplified EV Type and CAFV Status

df["EV Type"] = df["Electric Vehicle Type"].map({
    "Battery Electric Vehicle (BEV)": "BEV",
    "Plug-in Hybrid Electric Vehicle (PHEV)": "PHEV"
})

df["EV Type"] = df["EV Type"].fillna("Unknown")


df["CAFV Status"] = df[
    "Clean Alternative Fuel Vehicle (CAFV) Eligibility"
].map({
    "Clean Alternative Fuel Vehicle Eligible": "Eligible",
    "Eligibility unknown as battery range has not been researched": "Unknown",
    "Not eligible due to low battery range": "Not Eligible"
})

df["CAFV Status"] = df["CAFV Status"].fillna("Unknown")


print("EV Type values:")
print(df["EV Type"].value_counts())

print("\nCAFV Status values:")
print(df["CAFV Status"].value_counts())
```

```python
# Step 6: Clean Electric Range

import numpy as np

# Convert to numeric
df["Electric Range"] = pd.to_numeric(
    df["Electric Range"],
    errors="coerce"
)

# Replace 0 with missing value
df["Electric Range Clean"] = df["Electric Range"].replace(0, np.nan)

print("Original Electric Range:")
print(df["Electric Range"].describe())

print("\nClean Electric Range:")
print(df["Electric Range Clean"].describe())

print("\nMissing values after cleaning:",
      df["Electric Range Clean"].isna().sum())
```

```python
# Step 7: Clean Base MSRP

df["Base MSRP"] = pd.to_numeric(
    df["Base MSRP"],
    errors="coerce"
)

# Replace 0 with missing value
df["MSRP Clean"] = df["Base MSRP"].replace(0, np.nan)

print("Original MSRP:")
print(df["Base MSRP"].describe())

print("\nClean MSRP:")
print(df["MSRP Clean"].describe())

print("\nMissing values after cleaning:",
      df["MSRP Clean"].isna().sum())
```

```python
# Step 8: Create analytical columns

# Vehicle age
current_year = 2026

df["Vehicle Age"] = current_year - df["Model Year"]

# Remove unrealistic future model years
df.loc[df["Vehicle Age"] < 0, "Vehicle Age"] = np.nan


# Range category
df["Range Category"] = pd.cut(
    df["Electric Range Clean"],
    bins=[-1, 100, 200, 300, np.inf],
    labels=[
        "Low Range",
        "Medium Range",
        "High Range",
        "Very High Range"
    ]
)


# Check results
print("Vehicle Age:")
print(df["Vehicle Age"].describe())

print("\nRange Category:")
print(df["Range Category"].value_counts(dropna=False))
```

```python
# Step 9: Save the cleaned dataset

output_file = "EV_Population_Cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("File:", output_file)
```
